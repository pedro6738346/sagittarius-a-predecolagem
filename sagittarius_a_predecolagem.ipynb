ANÁLISE RELATÓRIO PRÉ - DECOLAGEM

```python
##Relatório TelemetriaSIMULAÇÃO 1:


system = {
    "temperatura_interna": 22,
    "temperatura_externa": -30,
    "integridade_estrutural": 1,
    "nivel_energia": 91,
    "pressao_tanque_principal": 320,
    "status_propulsao": 1,
    "status_navegacao": 1,
    "status_suporte_vida": 1,
    "status_comunicacao": 1
}

erros_encontrados = []

if system["temperatura_interna"] < 18:
  erros_encontrados.append("Temperatura Interna Baixa")
elif system["temperatura_interna"] > 27:
  erros_encontrados.append("Temperatura Interna Alta")

if system["temperatura_externa"] < -60:
  erros_encontrados.append("Temperatura Externa Baixa")
elif system["temperatura_externa"] > 60:
  erros_encontrados.append("Temperatura Externa Alta")

if system["integridade_estrutural"] != 1:
  erros_encontrados.append("Integridade Estrutural Comprometida")

if system["nivel_energia"] < 85:
  erros_encontrados.append("Nível de Energia Baixo")

if system["pressao_tanque_principal"] < 300:
  erros_encontrados.append("Pressão do Tanque Principal Baixa")
elif system["pressao_tanque_principal"] > 350:
  erros_encontrados.append("Pressão do Tanque Principal Alta")

if system["status_propulsao"] != 1:
  erros_encontrados.append("Propulsão Desligada")

if system["status_navegacao"] != 1:
  erros_encontrados.append("Navegação Desligada")

if system["status_suporte_vida"] != 1:
  erros_encontrados.append("Suporte de Vida Desligado")

if system["status_comunicacao"] != 1:
  erros_encontrados.append("Comunicação Desligada")


if len(erros_encontrados) == 0:
    print("DECOLAGEM AUTORIZADA")
else:
    print(f"FALHA NA DECOLAGEM ANOMALIAS IDENTIFICADAS NO SISTEMA: {erros_encontrados}")

```

ANÁLISE ENERGÉTICA RELATÓRIO PRÉ-DECOLAGEM

```python
#Análise Energética - Simulação 1

system = {
    "capacidade_total_bateria": 450,
    "porcentagem_energia_atual": 91,
    "consumo_estimado_apos_decolagem": 120,
    "perda_energetica_estimada": 8
}


energia_disponivel = (system["capacidade_total_bateria"] * system["porcentagem_energia_atual"])/100
porcentagem_que_sobrou_apos_perda = 100 - system["perda_energetica_estimada"]
apos_perda = (energia_disponivel * porcentagem_que_sobrou_apos_perda)/100
apos_decolagem = round(apos_perda - system["consumo_estimado_apos_decolagem"],2)
porcentagem_equivalente_capacidade_total = round((apos_decolagem/system["capacidade_total_bateria"]) * 100, 2)



if system["porcentagem_energia_atual"] < 85:
  print(f"DECOLAGEM CANCELADA! ENERGIA BAIXA: \n Energia disponível antes da decolagem:{energia_disponivel} kWh \n Energia após perdas e decolagem: {apos_decolagem} kWh \n Porcentagem restante comparado a Capacidade Total: {porcentagem_equivalente_capacidade_total}%")
else:
   print(f"DECOLAGEM AUTORIZADA! \n Energia disponível antes da decolagem:{energia_disponivel} kWh \n Energia após perdas e decolagem: {apos_decolagem} kWh \n Porcentagem restante comparado a Capacidade Total: {porcentagem_equivalente_capacidade_total}%")


```
